```python
import pandas as pd
from IPython.display import display
from sklearn.datasets import load_breast_cancer

breast_cancer_data = load_breast_cancer(as_frame=True)
breast_cancer_df = breast_cancer_data.frame.copy()
breast_cancer_df['diagnosis'] = breast_cancer_df['target'].map({0: 'Malignant', 1: 'Benign'})
print(breast_cancer_df.head())
```

```python
import seaborn as sns
import pandas as pd

# in python sns there are some dataset already existing, we can load them using sns.load_dataset('dataset_name')
# The breast cancer dataset is available from sklearn instead of seaborn.
df= breast_cancer_df.copy()
print(df.head())
```

```python
print(df.describe().loc['mean', 'mean radius'])
print(df['mean radius'].mean())
```

```python
#6 Are there any duplicate values? 
print(df.duplicated().sum())
```

```python
#remove all duplicated values

df=df.drop_duplicates()
df.reset_index(drop=True, inplace=True)
df.duplicated().sum()
```

```python
#7 How is the correlation among columns?

#df.select_dtypes(include=['float64', 'int64']).corr()
#non column correlation
df.select_dtypes(include=['float64', 'int64']).corrwith(df['mean radius'])
```

```python
import numpy as np
df.select_dtypes(include=[np.number]).corr()['target']
```

```python
import matplotlib.pyplot as plt

numeric_df= df.select_dtypes(include='number')
correlations= numeric_df.corr(method='pearson')

plt.figure(figsize=(10,8))
sns.heatmap(correlations, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)

plt.title('Correlation Heatmap (Numeric Features Only)')
plt.show()
```

```python
sns.countplot(df, x='target', color='steelblue')
```

```python
df['target'].value_counts()
```

```python
df['diagnosis'].value_counts().plot(kind='pie', autopct='%.2f')
plt.title('Distribution of Tumors by Diagnosis')
plt.show()
```

```python
plt.hist(df['mean radius'], bins=30)
plt.xlabel('Mean Radius')
plt.ylabel('Frequency')
plt.title('Distribution of Tumors by Mean Radius')
plt.show()
```

```python
sns.displot(df['mean radius'], kde=True)
plt.xlabel('Mean Radius')
plt.ylabel('Frequency')
plt.title('Distribution of Tumors by Mean Radius')
plt.show()
```

```python
sns.boxplot(x='mean radius', data=df)
```

```python
sns.kdeplot(df['mean radius'], shade=True)
plt.xlabel('Mean Radius')
plt.ylabel('Density')
plt.title('Kernel Density Estimate of Tumors by Mean Radius')
plt.show()
```
