```python
# Relational Operators
a = 10
b = 20

print(f"a = {a}, b = {b}")
print(f"Equal: a == b -> {a == b}")
print(f"Not Equal: a != b -> {a != b}")
print(f"Greater Than: a > b -> {a > b}")
print(f"Less Than: a < b -> {a < b}")
print(f"Greater or Equal: a >= b -> {a >= b}")
print(f"Less or Equal: a <= b -> {a <= b}")
```

## Part 9: Data Structures

Data structures are essential for organizing and managing data in data mining applications.


```python
# Lists - Mutable sequences
numbers = [11, 21, 31, 41, 51]
print(f"Initial list: {numbers}")

# Adding elements
numbers.append(6) # Entire object as a single element

print(f"After append(6): {numbers}")
numbers.insert(2, 10)
print(f"After insert(2, 10): {numbers}")
numbers.extend([7, 8, 9]) # Elements are added one by one
print(f"After extend([7, 8, 9]): {numbers}")
```

```python
# Removing elements
numbers = [1, 2, 3, 5, 5, 3, 6]
print(f"Original: {numbers}")

numbers.remove(5)  # Removes first occurrence
print(f"After remove(5): {numbers}")

while 5 in numbers:
    numbers.remove(5) # remove all occurance of value 5

popped = numbers.pop()  # Removes last element
print(f"Popped: {popped}, List: {numbers}")

popped = numbers.pop(2)  # Removes index 2
print(f"Popped index 2: {popped}, List: {numbers}")

del numbers[1]
print(f"After del numbers[1]: {numbers}")

numbers.clear()
print(f"After clear(): {numbers}")
```

```python
# List sorting and reversing
numbers = [5, 2, 8, 1, 9, 3]
print(f"Original: {numbers}")

numbers.sort()
print(f"After sort(): {numbers}")

numbers.sort(reverse=True)
print(f"After sort(reverse=True): {numbers}")

numbers.reverse()
print(f"After reverse(): {numbers}")

# Sorted without modifying
numbers = [5, 2, 8, 1, 9, 3]
sorted_list = sorted(numbers)
print(f"Original: {numbers}")
print(f"Sorted: {sorted_list}")
```

```python
# List operations
numbers = [1, 2, 3, 4, 5]
print(f"List: {numbers}")
print(f"Length: {len(numbers)}")
print(f"Max: {max(numbers)}")
print(f"Min: {min(numbers)}")
del sum
# if you have used a variable same as function name
print(f"Sum: {sum(numbers)}")
print(f"Average: {sum(numbers) / len(numbers)}")
print(f"Index of 3: {numbers.index(3)}")
print(f"Count of 2: {numbers.count(2)}")
```

```python
# List comprehension
squares = [x**2 for x in range(10)]
print(f"Squares: {squares}")

even_numbers = [x for x in range(20) if x % 2 == 0]
print(f"Even numbers: {even_numbers}")

# Nested list comprehension
matrix = [[j for j in range(5)] for i in range(3)]
print(f"Matrix:")
for row in matrix:
    print(row)
```

```python
# Tuples are immutable, which means their elements cannot be changed after creation.
t = (1, 2, 3, 4, 5)
print(f"Tuple: {t}")
print(f"Type: {type(t)}")
print(f"Length: {len(t)}")
print(f"Index 2: {t[2]}")
print(f"Slice 1-3: {t[1:4]}")

# Tuple unpacking
a, b, c = (1, 2, 3)
print(f"a={a}, b={b}, c={c}")

# Single element tuple
single = (5,)
print(f"Single element tuple: {single}")

# You can't add, delete elments of a tuple directly. You can only create a new tuple.
t = (1, 2, 3)
t = t + (4,)
print(t)
t = t[:1] + (10,) + t[1:]
print(t)
```

```python
# Sets - Unordered collections of unique elements
s = {1, 2, 3, 4, 5}
print(f"Set: {s}")

s.add(6)
print(f"After add(6): {s}")
s.add(3)  # No duplicate
print(f"After add(3): {s}")

s.remove(4)
print(f"After remove(4): {s}")

# Set operations
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
print(f"Set1: {set1}")
print(f"Set2: {set2}")
print(f"Union: {set1.union(set2)}")
print(f"Intersection: {set1.intersection(set2)}")
print(f"Difference (set1 - set2): {set1.difference(set2)}")
print(f"Symmetric difference: {set1.symmetric_difference(set2)}")
```

## Part 10: Functions

Functions help organize code and promote reusability.


```python
# Basic function
def add(a, b):
    return a + b

# Function with multiple returns
def calculate(x, y):
    return x + y, x - y, x * y, x / y

# Using functions
result = add(10, 20)
print(f"add(10, 20) = {result}")

sum_, diff, prod, quot = calculate(10, 5)
print(f"calculate(10, 5): sum={sum_}, diff={diff}, prod={prod}, quot={quot:.2f}")
```

```python
# Default arguments
def power(x, p=2):
    return x ** p

print(f"power(5) = {power(5)}")  # Uses default p=2
print(f"power(5, 3) = {power(5, 3)}")  # Uses p=3
print(f"power(2, 4) = {power(2, 4)}")
```

```python
# Keyword arguments
def student_info(name, age, cgpa=0.0):
    print(f"Name: {name}, Age: {age}, CGPA: {cgpa}")

student_info("Rahim", 22, 3.75)
student_info(name="Sadia", age=23)
student_info(cgpa=3.90, name="Karim", age=21)
```

```python
# Variable number of arguments
def average(*args):
    if len(args) == 0:
        return 0
    return sum(args) / len(args)

print(f"average(1, 2, 3) = {average(1, 2, 3)}")
print(f"average(1, 2, 3, 4, 5) = {average(1, 2, 3, 4, 5)}")
print(f"average() = {average()}")

def print_students(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_students(Rahim=85, Karim=78, Sadia=92)
```

```python
# Lambda functions
square = lambda x: x * x
print(f"square(5) = {square(5)}")

add = lambda a, b: a + b
print(f"add(10, 20) = {add(10, 20)}")

# Using lambda with map
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
print(f"Squared numbers: {squared}")

# Using lambda with filter
even = list(filter(lambda x: x % 2 == 0, numbers))
print(f"Even numbers: {even}")
```

```python
# Docstrings
def calculate_average(marks):
    """
    Calculate average of a list of marks.
    
    Args:
        marks: List of numeric values
    
    Returns:
        Average of the marks
    """
    if not marks:
        return 0
    return sum(marks) / len(marks)

print(f"Docstring: {calculate_average.__doc__}")
print(f"calculate_average([85, 90, 78]) = {calculate_average([85, 90, 78]):.2f}")
```

### Question 3 (5 Marks)

Input an integer N and print:
1 2 3 ... N
Also print the summation from 1 to N.


```python
# Solution for Question 3
N = int(input("Enter a positive integer N: "))

if N > 0:
    # Print sequence
    for i in range(1, N + 1):
        print(i, end=" " if i < N else "\n")
    
    # Calculate sum using formula
    total = N * (N + 1) // 2
    print(f"Sum from 1 to {N}: {total}")
    
    # Alternative using loop
    sum_loop = 0
    for i in range(1, N + 1):
        sum_loop += i
    print(f"Sum from 1 to {N} (using loop): {sum_loop}")
else:
    print("Please enter a positive integer")
```

2. Smallest number
3. Average value


```python
# Solution for Question 4
numbers = []
print("Enter 3 numbers:")
for i in range(3):
    num = float(input(f"Enter number {i+1}: "))
    numbers.append(num)

print(f"\nList: {numbers}")
print(f"Largest: {max(numbers)}")
print(f"Smallest: {min(numbers)}")
print(f"Average: {sum(numbers) / len(numbers):.2f}")
```

### Question 5 (10 Marks)

Input marks of N students and store them in a dictionary.

Display:

1. Highest scorer name and marks
2. Average marks
3. Number of students obtaining marks ≥ 80


```python
# Solution for Question 5
students = {}
n = int(input("Enter number of students: "))

for i in range(n):
    name = input(f"Enter name of student {i+1}: ")
    marks = float(input(f"Enter marks for {name}: "))
    students[name] = marks

print(f"\nStudent Data: {students}")

# Find highest scorer
if students:
    highest_name = max(students, key=students.get)
    highest_marks = students[highest_name]
    print(f"\nHighest scorer: {highest_name} with {highest_marks} marks")
    
    # Calculate average
    avg = sum(students.values()) / len(students)
    print(f"Average marks: {avg:.2f}")
    
    # Count students with marks >= 80
    high_achievers = [name for name, marks in students.items() if marks >= 80]
    print(f"Number of students with marks >= 80: {len(high_achievers)}")
    if high_achievers:
        print(f"Students with >= 80: {', '.join(high_achievers)}")
else:
    print("No student data available")
```

### Bonus Question (5 Marks)

Input a sentence and count the frequency of each word.

Example:
"I love data mining and I love python"

Output: {'I':2, 'love':2, 'data':1, 'mining':1, 'and':1, 'python':1}


```python
# Solution for Bonus Question
sentence = input("Enter a sentence: ")
print(f"Original: {sentence}")

# Split into words and count
words = sentence.split()
word_count = {}

for word in words:
    # Remove punctuation and convert to lowercase for better counting
    clean_word = word.strip('.,!?;:()"\'').lower()
    if clean_word:
        word_count[clean_word] = word_count.get(clean_word, 0) + 1

print(f"\nWord frequency:")
for word, count in word_count.items():
    print(f"{word}: {count}")

print(f"\nDictionary output: {word_count}")
```
