# Lab 4 – Data Preprocessing

## Cell 1: Import all libraries

```python
import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder, LabelEncoder
```

## Cell 2: Load dataset and display dataset before preprocessing

```python
df = pd.read_excel("dataset_lab5.xlsx")
df
```

## Cell 3: Identify missing values

```python
df.isnull().sum()
```

## Cell 4: Create a copy of the dataset for preprocessing

```python
df_full = df.copy()
df_full
```

## Cell 5: Replace missing Age value with mean

```python
df_full["Age"] = df_full["Age"].fillna(df_full["Age"].mean())
df_full["Age"].isnull().sum()
```

## Cell 6: Replace missing Income value with mean

```python
df_full["Income"] = df_full["Income"].fillna(df_full["Income"].mean())
df_full["Income"].isnull().sum()
```

## Cell 7: Replace missing Gender value with mode

```python
df_full["Gender"] = df_full["Gender"].fillna(df_full["Gender"].mode()[0])
df_full["Gender"].isnull().sum()
```

## Cell 8: Replace missing City value with mode

```python
df_full["City"] = df_full["City"].fillna(df_full["City"].mode()[0])
df_full["City"].isnull().sum()
```

## Cell 9: Confirm no missing values

```python
df_full.isnull().sum()
```

## Cell 10: Separate numeric and categorical columns

```python
num_cols = df_full.select_dtypes(include=np.number).columns
cat_cols = df_full.select_dtypes(include=["object", "category", "string"]).columns
num_cols, cat_cols
```

## Cell 11: Standard Scaling

```python
scaler = StandardScaler()
scaled = scaler.fit_transform(df_full[num_cols])
scaled_df = pd.DataFrame(scaled, columns=num_cols)
scaled_df.head()
```

## Cell 12: Min-Max Scaling

```python
Minmax = MinMaxScaler()
norm = Minmax.fit_transform(df_full[num_cols])
norm_df = pd.DataFrame(norm, columns=num_cols)
norm_df.head()
```

## Cell 13: Apply Label Encoding to Gender

```python
le = LabelEncoder()
df_full["Gender"] = le.fit_transform(df_full["Gender"])
df_full
```

## Cell 14: Apply One-Hot Encoding to City

```python
ohe = OneHotEncoder(drop="first", sparse_output=False, handle_unknown="ignore", dtype=np.int32)
encoded = ohe.fit_transform(df_full[["City"]])
encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(["City"]))

display(df_full[["City"]].head())
display(encoded_df.head())
```

## Cell 15: Final dataset using Min-Max Scaling

```python
df_minmax = pd.concat([norm_df, df_full[["Gender"]], encoded_df], axis=1)
df_minmax
```

## Cell 16: Final dataset using Standard Scaling

```python
df_standard = pd.concat([scaled_df, df_full[["Gender"]], encoded_df], axis=1)
df_standard
```

## Cell 17: Display dataset before and after preprocessing

```python
print("Original DataFrame: ")
display(df)

print("\nAfter Preprocessing with Min-Max Scaling: ")
display(df_minmax)

print("\nAfter Preprocessing with Standard Scaling: ")
display(df_standard)
```
