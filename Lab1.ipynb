```python
# Arithmetic - for feature engineering

x, y=10,3
print(f"Division: {x/y}")
print(f"Floor division : {x//y}")
print(f"Modulus: {x%y}")
print(f"Power: {x**y}")

#comparison - for filtering

print(f"x >y: {x>y}")
print(f"x == y: {x==y}")

# Logical - for conditions

age= 22
print(f"Age>= 18 and Age <=60: {age>=18 and age<=60}")
```

```python
# if-elif-else -for data categorization

marks = 85
if marks>=80:
    grade = 'A'
elif marks>=60:
    grade = 'B'
else:
    grade= 'C'

print(f"Marks: {marks}, Grade: {grade}")
```

```python
#while loop - until condition met 
total =0
count=0
print ("Enter scores (enter -1 to stop):")

while True: 
    score=float(input())
    if score==-1:
        break
    total +=score
    count+=1
if count>0:
    print(f"Avarage: {total/count:.2f}")
```

```python
# string 

text= "Data Mining and Machine Learning"

#common string operations
print(f"Length: {len(text)}")
print(f"Uppercase: {text.upper()}")
print(f"Lowercase: {text.lower()}")

#splitting - essential for text proccessing 
words= text.split()
print(f"Words: {words}")
```

```python
#searching 
print(f"Contains 'Mining': {'Mining' in text}")
print(f"Position of 'Mining': {text.find('Mining')}")

#Replacing

new_text=text.replace('Mining', 'Analysis')
print(f"Replaced: {new_text}")
```

```python
#creating lists

data=[85, 92, 90, 78, 98, 120]

#Basic operations
print(f"Data: {data}")
print(f"Length: {len(data)}")
print(f"Max: {max(data)}")
print(f"Min: {min(data)}")
print(f"Sum: {sum(data)}")
print(f"Average: {sum(data)/len(data):.2f}")
```

```python
# Adding and removing elements
data=[10, 20 , 30 , 40, 50]
print(f"Original Data: {data}")

data.append(60)
print(f"After append: {data}")

data.insert(0, 25)
print(f"After insert: {data}")

removed= data.pop() #remove last
print(f"Removed: {removed}, After pop: {data}")
removed=data.pop(2)
print(f"Removed: {removed}, After pop: {data}")
```

```python
#sorting -essential for data analysis

scores= [99, 88, 75, 85, 90]
scores.sort()
print(f"Sorted ascending: {scores}")
scores.sort(reverse=True)
print(f"Sorted decending: {scores}")

#list comprehension - efficient data transformation 
original =[1,2,3,4,5]
squared=[x**2 for x in original]
print(f"Original : {original}")
print(f"Squared: {squared}")

even= [x for x in range(10) if x%2 ==0]
print(f"Even numbers: {even}")
```
