```python
import seaborn as sns
import pandas as pd

# in python sns there are some dataset already existing, we can load them using sns.load_dataset('dataset_name')
df= sns.load_dataset('titanic')
print(df.head())
```

```python
#1 What is the size of the dataset? 

print("Dataset shape: ", df.shape)
print("Number of row: ", df.shape[0])
print("Number of column: ", df.shape[1])

print("Dimention of the dataset: ", df.ndim)
```

```python
#2 How does the dataset look like?
print(df.head())
print(df.tail(10))
```

```python
#5 How does the dataset look mathematically? (mean, median, mode, standard deviation, etc.)

display(df.describe())
```

```python
print(df.describe().loc['mean', 'age'])
print(df['age'].mean())
```

```python
#6 Are there any duplicate values? 
print(df.duplicated().sum())
```

```python
#remove all duplicated values

df=df.drop_duplicates()
df.reset_index(drop=True, inplace=True)
df.duplicated().sum()
```

```python
#7 How is the correlation among columns?

#df.select_dtypes(include=['float64', 'int64']).corr()
#non column correlation
df.select_dtypes(include=['float64', 'int64']).corrwith(df['age'])
```

```python
import numpy as np
df.select_dtypes(include=[np.number]).corr()['survived']
```

```python
import matplotlib.pyplot as plt

numeric_df= df.select_dtypes(include='number')
correlations= numeric_df.corr(method='pearson')

plt.figure(figsize=(10,8))
sns.heatmap(correlations, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)

plt.title('Correlation Heatmap (Numeric Features Only)')
plt.show()
```

```python
sns.countplot(df, x=df['survived'])
```

```python
df['survived'].value_counts()
```

```python
df['sex'].value_counts().plot(kind='pie', autopct='%.2f')
plt.title('Distribution of Passengers by Sex')
plt.show()
```

```python
plt.hist(df['age'], bins=30)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Distribution of Passengers by Age')
plt.show()
```

```python
sns.displot(df['age'], kde=True)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Distribution of Passengers by Age')
plt.show()
```

```python
sns.boxplot(x='age', data=df)
```

```python
sns.kdeplot(df['age'], shade=True)
plt.xlabel('Age')
plt.ylabel('Density')
plt.title('Kernel Density Estimate of Passengers by Age')
plt.show()
```
