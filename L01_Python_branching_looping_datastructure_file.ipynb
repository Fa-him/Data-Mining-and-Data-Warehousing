## 3. Operators

Operators used in data manipulation:

## 4. Branching (Conditions)

Conditions for data filtering and categorization:

```python
# if-elif-else - for data categorization
marks = 85

if marks >= 80:
    grade = 'A'
elif marks >= 60:
    grade = 'B'
else:
    grade = 'C'
print(f"Marks: {marks}, Grade: {grade}")

# Categorizing data
temperature = 35
if temperature > 30:
    category = 'Hot'
elif temperature > 20:
    category = 'Warm'
else:
    category = 'Cold'
print(f"Temperature: {temperature}C, Category: {category}")
```

## 5. Loops

Iterating through data:

```python
# for loop - iterating through data
data = [10, 20, 30, 40, 50]
for value in data:
    print(value, end=" ")
print()

# range() - for numerical ranges
for i in range(1, 6):
    print(f"Square of {i}: {i**2}")
```

```python
# while loop - until condition met
total = 0
count = 0
print("Enter scores (enter -1 to stop):")
while True:
    score = float(input())
    if score == -1:
        break
    total += score
    count += 1
if count > 0:
    print(f"Average: {total/count:.2f}")
```

```python
# break and continue - data filtering
data = [1, -5, 3, 0, -2, 7]
positive = []
for num in data:
    if num < 0:
        continue  # Skip negative
    if num == 0:
        break     # Stop at zero
    positive.append(num)
print(f"Positive values before first zero: {positive}")
```

## 7. Lists - Data Collections

Lists are fundamental for data storage and manipulation:

```python
# Creating lists
data = [85, 92, 78, 90, 88]

# Basic operations
print(f"Data: {data}")
print(f"Length: {len(data)}")
print(f"Max: {max(data)}")
print(f"Min: {min(data)}")
print(f"Sum: {sum(data)}")
print(f"Average: {sum(data)/len(data):.2f}")
```

```python
# Adding and removing elements
data = [10, 20, 30, 40]
print(f"Original: {data}")

data.append(50)        # Add at end
print(f"After append: {data}")

data.insert(0, 25)     # Insert at index
print(f"After insert: {data}")

removed = data.pop()   # Remove last
print(f"Removed: {removed}, After pop: {data}")

removed = data.pop(1)   # Remove index 1
print(f"Removed: {removed}, After pop: {data}")

data.remove(25)        # Remove specific value
print(f"After remove: {data}")
```

## 10. Sets - Unique Values

Sets are useful for unique value identification:

```python
# Removing duplicates
duplicates = [1, 2, 2, 3, 3, 3, 4, 5, 5]
unique = set(duplicates)
print(f"Original: {duplicates}")
print(f"Unique: {unique}")

# Set operations
set1 = {1, 2, 2, 3, 4, 5} # duplicate values are automatically removed
print(f"{set1=}")
set2 = {4, 5, 6, 7, 8}
print(f"{set2=}")

print(f"Union: {set1.union(set2)}")
print(f"Intersection: {set1.intersection(set2)}")
print(f"Difference: {set1.difference(set2)}")
```

## 11. Functions - Reusable Code

Functions for modular data processing:

```python
# Basic function
def calculate_average(data):
    if len(data) == 0:
        return 0
    return sum(data) / len(data)

scores = [85, 92, 78, 90]
avg = calculate_average(scores)
print(f"Average: {avg:.2f}")
```

```python
# Function with multiple operations
def analyze_data(data):
    if not data:
        return None
    return {
        'mean': sum(data) / len(data),
        'max': max(data),
        'min': min(data),
        'length': len(data)
    }

numbers = [10, 20, 30, 40, 50]
stats = analyze_data(numbers)
for key, value in stats.items():
    print(f"{key}: {value}")
```

```python
# Lambda functions
square = lambda x: x**2
print(f"Square of 5: {square(5)}")

# Using lambda with map
data = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, data))
print(f"Squared: {squared}")

# Using lambda with filter
even = list(filter(lambda x: x % 2 == 0, data))
print(f"Even: {even}")
```

## 12. File Handling - Reading Data

Reading data files is essential for data mining:

```python
# Writing data to file
data = ["Name,Age,Score", "Rahim,22,85", "Karim,23,78", "Sadia,21,92"]
with open('students.csv', 'w') as file:
    for line in data:
        file.write(line + '\n')
print("File created successfully")
```

```python
# Reading CSV data
students = []
with open('students.csv', 'r') as file:
    headers = file.readline().strip().split(',')
    for line in file:
        values = line.strip().split(',')
        student = dict(zip(headers, values))
        students.append(student)

print("Student data:")
for student in students:
    print(student)
```

## 13. Useful Functions for Data Mining

```python
# Data statistics
def get_statistics(data):
    return {
        'mean': sum(data) / len(data),
        'max': max(data),
        'min': min(data),
        'range': max(data) - min(data),
        'sum': sum(data)
    }

# Data normalization (min-max scaling)
def normalize(data):
    min_val = min(data)
    max_val = max(data)
    if max_val == min_val:
        return [0] * len(data)
    return [(x - min_val) / (max_val - min_val) for x in data]

# Data validation
def clean_data(data, min_val=None, max_val=None):
    cleaned = []
    for x in data:
        if min_val is not None and x < min_val:
            continue
        if max_val is not None and x > max_val:
            continue
        cleaned.append(x)
    return cleaned

# Example usage
scores = [85, 92, 78, 90, 88, 95, 82]
print(f"Original: {scores}")
print(f"Statistics: {get_statistics(scores)}")
print(f"Normalized: {normalize(scores)}")
print(f"Cleaned (80-90): {clean_data(scores, 80, 90)}")
```
