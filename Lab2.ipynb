````python
import numpy as np
array = np.random.randint(0,10,(3,4))  # 3x4 array of random integers from 0 to 9
print(array)
````

````python
print("Array min: ", array.min())
print("Array max: ", array.max())
print("Array mean: ", array.mean())
print("Array variance: ", array.var())
print("Array standard deviation: ", array.std())
````

````python
import pandas as pd
data={
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [25, 30, 35, 40, 45],
    "Score": [85, 90, 95, 80, 75]
}
df= pd.DataFrame(data)
df
````

````python
result_and =df.query('Age > 20 and Score <= 80')
print("Result of AND operation:\n", result_and)
````

````python
result_string = df.query("Name=='Charlie' or Name=='Eva'")
print("Charlie or Eva\n", result_string)
````
