# Lab 5 Evaluation

## Cell 1: Import all libraries

```python
import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder, LabelEncoder
```

## Cell 2: Load dataset

```python
df = pd.read_excel("dataset_lab5.xlsx")
df
```

## Cell 3: Identify missing values

```python
df.isnull().sum()
```

## Cell 4: Replace missing Age value with mean

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

## Cell 5: Replace missing Income value with mean

```python
df["Income"] = df["Income"].fillna(df["Income"].mean())
df
```

## Cell 6: Replace missing Gender value with mode

```python
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])
df
```

## Cell 7: Replace missing City value with mode

```python
df["City"] = df["City"].fillna(df["City"].mode()[0])
df
```

## Cell 8: Confirm no missing values

```python
df.isnull().sum()
```

## Cell 9: Apply Min-Max Scaling to Age and Income

```python
Minmax = MinMaxScaler()

df[["Age", "Income"]] = Minmax.fit_transform(df[["Age", "Income"]])

df
```

## Cell 10: Apply Label Encoding to Gender

```python
le = LabelEncoder()

df["Gender"] = le.fit_transform(df["Gender"])

df
```

## Cell 11: Apply One Hot Encoding to City

```python
ohe = OneHotEncoder(sparse_output=False, dtype=np.int32)

encoded = ohe.fit_transform(df[["City"]])

city_encoded = pd.DataFrame(
    encoded,
    columns=ohe.get_feature_names_out(["City"])
)

city_encoded
```

## Cell 12: Combine One Hot Encoded City with dataset

```python
df = pd.concat([df.drop("City", axis=1), city_encoded], axis=1)

df
```
